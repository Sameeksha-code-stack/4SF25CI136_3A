```python
import pandas as pd
import numpy as np
```

```python
arr = np.random.randint(1, 11, size=(5, 3))

df = pd.DataFrame(arr, columns=["A", "B", "C"])

print("Original DataFrame:")
print(df)

df["Product"] = df["A"] * df["B"]

print("After adding Product:")
print(df)
```

```python
arr = np.random.randint(1, 10, size=(4, 3))

df = pd.DataFrame(arr, columns=["A", "B", "C"])

print(df)

print("Row-wise sum:")
print(df.sum(axis=1))

print("Column-wise sum:")
print(df.sum(axis=0))
```

```python
random_data = np.random.randint(1, 21, size=(5, 3))
df = pd.DataFrame(random_data, columns=["A", "B", "C"])
df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
print("DataFrame with Nan values:")
print(df)
df_filled = df.fillna(df.mean())
print("DataFrame after filling NaN values with column mean:")
print(df_filled)
```

```python
random_data = np.random.randint(1, 21, size=(6, 4))
df = pd.DataFrame(random_data, columns=["W", "X", "Y", "Z"])
df.iloc[2, 1] = np.nan
df.iloc[4, 3] = np.nan
print("DataFrame with NaN values:")
print(df)
df_drop=df.dropna()
print("DataFrame after dropping rows with NaN values:")
print(df_drop)
```

```python
rows=10
categories=np.random.choice(['A', 'B', 'C'], size=rows)
values=np.random.randint(10, 100, size=rows)
df=pd.DataFrame({'Category': categories, 'Value': values})
print("Original DataFrame:")
print(df)
grouped=df.groupby('Category')['Value'].agg(["sum", "mean"])
print("Sum and Mean:")
print(grouped)
```

```python
df=pd.DataFrame({
    "Product": ["Pen", "Book", "Pencil", "Bag", "Bottle", "Notebook"],
    "Category": ["A", "B", "A", "C", "C", "B"],
    "Sales": [100, 200, 150, 300, 250, 180]
})

print(df)

result = df.groupby("Category")["Sales"].sum()

print("Total Sales:")
print(result)
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 85]
})

print("DataFrame 1:")
print(df1)

print("DataFrame 2:")
print(df2)

result = pd.merge(df1, df2, on="ID")

print("Merged DataFrame:")
print(result)
```

```python
index = [
    ["A", "A", "B", "B"],
    [1, 2, 1, 2]
]

df = pd.DataFrame(
    {"Value": [10, 20, 30, 40]},
    index=index
)

df.index.names = ["Category", "Number"]

print("DataFrame:")
print(df)

print(df.loc[("A",2):("B",1)])
```

```python
index=[
    ["A", "A", "B", "B"],
    [1, 2, 1, 2]
]   
df=pd.DataFrame({"value":np.random.randint(1, 100, size=4)}, index=index)
df.index.names=["Category", "SubCategory"]
print("DataFrame:")
print(df)
print("Sum of values in each category:")
print(df.groupby("Category").sum())
print("Sum of values for each subcategory:")
print(df.groupby("SubCategory").sum())
```

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(5, 3)), columns=["A", "B", "C"])

print("Original DataFrame:")
print(df)
def double_values(dataframe):
    return dataframe *2
df_doubled = df.map(double_values)
print("DataFrame after doubling values:")
print(df_doubled)
```

```python
df = pd.DataFrame(np.random.randint(1, 50, size=(6, 3)), columns=["A", "B", "C"])

print("DataFrame:")
print(df)

df["Sum"] = df.apply(lambda row: row["A"] + row["B"] + row["C"], axis=1)
print("DataFrame after adding Sum column:")
print(df)
```
